2. Напишите запрос, который для каждого региона выведет количество работодателей и вакансий в нём.
Среди регионов, в которых нет вакансий, найдите тот, в котором наибольшее количество работодателей.


```python
# текст запроса
query_5_2 = f'''
    WITH vac_counts AS (
        SELECT area_id, COUNT(id) AS vacancies_count
        FROM public.vacancies
        GROUP BY area_id
    ),
    emp_counts AS (
        SELECT area, COUNT(id) AS employers_count
        FROM public.employers
        GROUP BY area
    )

    SELECT
        a.name AS area_name,
        COALESCE(v.vacancies_count, 0) AS vacancies_count,
        COALESCE(e.employers_count, 0) AS employers_count
    FROM public.areas a
    LEFT JOIN vac_counts v ON a.id = v.area_id
    LEFT JOIN emp_counts e ON a.id = e.area
    ORDER BY vacancies_count ASC, employers_count DESC
'''
```

```python
# результат запроса
df_5_2 = pd.read_sql_query(query_5_2, connection)
df_5_2.head(10) # Выведем топ-10, чтобы сразу увидеть нужный регион с нулями вакансий
```

```python
# Дополнительное исследование: Кластеризация DS-вакансий
query_ = f'''
    WITH ds_vacancies AS (
        SELECT id, name
        FROM public.vacancies
        WHERE (
            name ILIKE '%data scientist%' OR
            name ILIKE '%data science%' OR
            name ILIKE '%исследователь данных%' OR
            (name LIKE '%ML%' AND name NOT ILIKE '%HTML%') OR
            name ILIKE '%machine learning%' OR
            name ILIKE '%машинн%обучен%'
        )
    )
    SELECT
        CASE
            WHEN name ILIKE '%computer vision%' OR name LIKE '%CV%' OR name ILIKE '%компьютерн%зрен%' THEN 'Computer Vision (CV)'
            WHEN name ILIKE '%nlp%' OR name ILIKE '%natural language%' OR name ILIKE '%текст%' THEN 'Natural Language Processing (NLP)'
            WHEN name ILIKE '%analyst%' OR name ILIKE '%аналитик%' THEN 'Data Analyst'
            WHEN name ILIKE '%engineer%' OR name ILIKE '%инженер%' OR name ILIKE '%разработчик%' THEN 'ML / Data Engineer'
            ELSE 'General Data Scientist / Researcher'
        END AS specialization,
        COUNT(id) AS count
    FROM ds_vacancies
    GROUP BY specialization
    ORDER BY count DESC
'''

# результат запроса
df_ = pd.read_sql_query(query_, connection)
df_
```
